```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

employees
 
```

```python
departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
departments
```

```python
pd.merge(employees, departments, on ="DeptID")  #INNER JOIN -no NAN
```

```python
pd.merge(employees, departments, on="DeptID", how="left")  #left join employees more prioritised
```

```python
pd.merge(employees, departments, on="DeptID", how="right")  #right join department more prioritised
```

```python
pd.merge(employees, departments, on="DeptID", how="outer")  #outer join --SHOWS ALL NAN
```
