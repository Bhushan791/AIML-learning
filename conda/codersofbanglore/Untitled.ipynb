```python
with open ("initialdata.txt", encoding="utf-8")  as file : 
    data = file.read() 
    
```

```python
data
```

```python
chunks = data.split("\n\n") 
```

```python
chunks= [ c for c in chunks if len(c)>3]
chunks[0]
```

```python
def clean_chunk(chunk): 
    chunk = chunk.strip()
    seperated_chunk = chunk.split('\n')  
    username = seperated_chunk[0]
    no_of_posts = seperated_chunk[1]
    no_of_followers = seperated_chunk[2]
    no_of_following = seperated_chunk[3]
    name= seperated_chunk[4]
    if(len(seperated_chunk)>5):
        type_of_page= seperated_chunk[5]
        bio= "\n".join(seperated_chunk[6:])
    else:
        type_of_page = "Unknown"
        bio= ""
    return {"username": username, "No. Of Posts": no_of_posts, "No. Of Followers": no_of_followers, "No of Following": no_of_following, "Name": name, "Type oF Page": type_of_page, "bio": bio}
```

```python

print(clean_chunk(chunks[4]))
```

```python
all_chunks = [] 
for chunk in chunks:
    cleaned_chunk = clean_chunk(chunk) 
    all_chunks.append(cleaned_chunk)
```

```python
print(all_chunks)
```
