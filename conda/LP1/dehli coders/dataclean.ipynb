```python
import json

#load data function
def clean_data(data):
    ##remove users with missing names--
    data['users'] = [user for user in data['users']    if user['name'].strip()]

    ##remove duplicate friends--
    for user in data['users']:
        ##convert the friends list to set and again convert back to the list
        user['friends'] = list(set(user['friends']))


    #remove inactive users
    data['users'] = [user for user in data['users'] if user['friends'] or user['liked_pages']]



    ##remove duplicate pages
    unique_pages={ }
    for page in data['pages']:
        unique_pages[page['id']] = page

    data['pages']= list(unique_pages.values())
    return data
    
    
    
data1 = json.load(open("data2.json"))
data2 = clean_data(data1)

json.dump(data2,  open("cleaned_data2.json", "w"), indent=4)
print("data has been cleaned successfully...201!!!")
```
