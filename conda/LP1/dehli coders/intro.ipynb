```python
import json
```

```python
##function to laod the data
def load_data(filename):
    with open (filename , "r") as f:
        data = json.load(f)
    return data
    
    
```

```python
data = load_data("data.json")
```

```python
#write a function to display users and their connections
def display_users(data):
    print("USERS AND THEIR CONNECTIONS:\n")
    
    # Create a quick lookup: user ID to user name
    id_to_name = {user['id']: user['name'] for user in data['users']}
    page_id_to_name= {page['id'] : page['name'] for page in data['pages']}
    
    
    for user in data['users']:
        # Get friend names from IDs
        friend_names = [id_to_name[friend_id] for friend_id in user['friends']]

        page_names = [page_id_to_name[page_id] for page_id in user['liked_pages']]
        
        print(f"ID - {user['id']} {user['name']} is friend with: {', '.join(friend_names)} and liked pages are {', '.join(page_names)}")
    
    print("\nPAGES INFORMATION:")
    for page in data['pages']:
        print(f"ID - {page['id']}: {page['name']}")

# Call the function
display_users(data)
```
