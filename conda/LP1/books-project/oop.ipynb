```python
class Employee:
    company = "HP"

    def __init__(self, name, salary):
        self.name =name
        self.salary = salary
        
    def details(self):
        print(f"The Company of {self.name} with salary {self.salary}  is {self.company}")


    @staticmethod
    def printtime():
        from datetime import datetime
        current_datetime = datetime.now()
        print(f"Current Date and Time: {current_datetime}")
        


e= Employee("harry", 34000)
e.details()
e.printtime()
```
