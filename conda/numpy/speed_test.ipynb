```python
# ##USING PYTHON LISTS


import time
size= 100000



l1= list(range(size))

l2= list(range(size))     

start= time.time()

add= [x+y for x,y in zip(l1,l2)]
end=time.time()
print(add[0:10])


print(end-start)

```

```python

##USING Numpy arrays

import numpy as np

size= 100000



l1= np.array(range(size))
l2= np.array(range(size))



start= time.time()

add= l1+l2
end=time.time()

print(add[0:10])


print(end-start)



```
