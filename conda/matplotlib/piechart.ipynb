```python
import matplotlib.pyplot as plt
plt.style.use("ggplot")
```

```python
labels  = [ "Sachin", "Shewag", "Kohli", "Yuvraj"]
runs = [10000, 12333, 19000, 8000] 
plt.title("Carer runs of indian batsmem")
plt.pie(runs,  labels=labels) 
plt.show()
```

```python
explode = [0.1, 0, 0, 0]  # Only highlight Sachin's slice

plt.pie(runs, labels=labels, explode=explode, shadow=True)
plt.title("Exploded Pie Example")
plt.show()
```
