```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedShuffleSplit
from matplotlib import pyplot
```

```python
df = pd.read_csv("housing.csv") 
    
```

```python
df['income_cat']  = pd.cut(df['median_income'],bins =[0,1.5, 3.0, 4.5, 6.0, np.inf], labels=[1,2,3,4,5]) 
```

```python

split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_idx, test_idx in split.split(df, df['income_cat']):
    strat_train_set= df.loc[train_idx]
    strat_test_set= df.loc[test_idx]
```

```python
## remove the income_Cat column
for sett in (strat_train_set, strat_test_set) :
    sett.drop("income_cat", axis=1, inplace=True)
```

```python
df1 = strat_train_set.copy()
```

```python
df1.plot(kind="scatter", x= "latitude", y="longitude", grid=True, cmap="jet", c="median_house_value")
```
