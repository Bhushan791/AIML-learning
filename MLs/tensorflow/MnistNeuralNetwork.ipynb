```python
import tensorflow as tf
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten,Input 
from tensorflow.keras.utils import to_categorical #used for one hot encoding 
from tensorflow.keras.datasets import mnist
```

```python
(x_train, y_train) , (x_test, y_test) =  mnist.load_data() 
```

```python
x_train = x_train/255.0
x_test = x_test/255.0   ##feature scaling teh pixels lie from 0-255 in mnist so we are ranging it 0-1 
```

```python
y_train = to_categorical(y_train, 10)
y_test= to_categorical(y_test,10)
```

```python
model =Sequential([
#Flatten(input_shape=(28,28)) ,
    Input(shape=(28,28)), 
    Flatten(), 
    Dense(128,activation="relu") , 
    Dense(10, activation="softmax")
])
```

```python
model.compile(
    optimizer = "adam", 
    loss="categorical_crossentropy" , 
    metrics = ["accuracy"]
)
```

```python
model.fit(x_train, y_train, epochs=5, batch_size=32)
```

```python
test_loss, test_acc= model.evaluate(x_test, y_test) 
print(f"Test Accuracy is {test_acc*100} % ")
```
